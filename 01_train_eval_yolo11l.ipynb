```python
from pathlib import Path

pred_dir = Path("/kaggle/working/sahi_preds")
pred_dir.mkdir(exist_ok=True)

(pred_dir / "labels").mkdir(exist_ok=True)
```

```python
from sahi.predict import get_sliced_prediction
from PIL import Image

for idx, img_path in enumerate(image_files):

    result = get_sliced_prediction(
        str(img_path),
        sahi_model,
        slice_height=512,
        slice_width=512,
        overlap_height_ratio=0.2,
        overlap_width_ratio=0.2,
    )

    img = Image.open(img_path)
    img_w, img_h = img.size

    label_file = pred_dir / "labels" / f"{img_path.stem}.txt"

    with open(label_file, "w") as f:

        for pred in result.object_prediction_list:

            x1 = pred.bbox.minx
            y1 = pred.bbox.miny
            x2 = pred.bbox.maxx
            y2 = pred.bbox.maxy

            xc = ((x1 + x2) / 2) / img_w
            yc = ((y1 + y2) / 2) / img_h

            bw = (x2 - x1) / img_w
            bh = (y2 - y1) / img_h

            conf = pred.score.value

            f.write(
                f"0 {xc} {yc} {bw} {bh} {conf}\n"
            )

    if (idx + 1) % 50 == 0:
        print(f"Processed {idx+1}/{len(image_files)}")
```

```python
from pathlib import Path
import numpy as np

pred_dir = Path("/kaggle/working/sahi_preds/labels")
gt_dir = Path(
    "/kaggle/input/datasets/abhinavsinha08/pothole-master-citylens2/Pothole_master/valid/labels"
)

def yolo_to_xyxy(box):
    xc, yc, w, h = box
    x1 = xc - w/2
    y1 = yc - h/2
    x2 = xc + w/2
    y2 = yc + h/2
    return np.array([x1,y1,x2,y2])

def iou(box1, box2):
    xA = max(box1[0], box2[0])
    yA = max(box1[1], box2[1])
    xB = min(box1[2], box2[2])
    yB = min(box1[3], box2[3])

    inter = max(0, xB-xA) * max(0, yB-yA)

    area1 = (box1[2]-box1[0])*(box1[3]-box1[1])
    area2 = (box2[2]-box2[0])*(box2[3]-box2[1])

    union = area1 + area2 - inter

    if union <= 0:
        return 0

    return inter / union

TP = 0
FP = 0
FN = 0

for gt_file in gt_dir.glob("*.txt"):

    pred_file = pred_dir / gt_file.name

    gt_boxes = []
    pred_boxes = []

    with open(gt_file) as f:
        for line in f:
            vals = line.strip().split()
            gt_boxes.append(
                yolo_to_xyxy(list(map(float, vals[1:5])))
            )

    if pred_file.exists():

        with open(pred_file) as f:
            for line in f:
                vals = line.strip().split()

                if len(vals) < 6:
                    continue

                pred_boxes.append(
                    yolo_to_xyxy(list(map(float, vals[1:5])))
                )

    matched_gt = set()

    for pred in pred_boxes:

        best_iou = 0
        best_idx = -1

        for idx, gt in enumerate(gt_boxes):

            if idx in matched_gt:
                continue

            current_iou = iou(pred, gt)

            if current_iou > best_iou:
                best_iou = current_iou
                best_idx = idx

        if best_iou >= 0.5:
            TP += 1
            matched_gt.add(best_idx)
        else:
            FP += 1

    FN += len(gt_boxes) - len(matched_gt)

precision = TP / (TP + FP + 1e-9)
recall = TP / (TP + FN + 1e-9)

print("TP:", TP)
print("FP:", FP)
print("FN:", FN)

print("\nSAHI Precision:", precision)
print("SAHI Recall:", recall)
```
